```python
import time
from transformers import AutoTokenizer, AutoModel

MODEL_NAME = 'distilbert-base-cased'

print(f'Загружаем {MODEL_NAME}...')
print('При первом запуске скачается ~260 MB (потом будет в кеше)')

start = time.time()

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME)

elapsed = time.time() - start

print(f"\n✅ Модель загружена за {elapsed:.1f} сек.")
print(f"Тип токенизатора: {type(tokenizer).__name__}")
print(f"Тип модели: {type(model).__name__}")
```
